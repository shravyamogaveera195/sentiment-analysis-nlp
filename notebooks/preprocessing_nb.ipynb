```python
import re
import nltk
import pandas as pd
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('stopwords')#multi language database with structured,highly frequent filler words
nltk.download('wordnet')# large,structured database and dictionary for english lang
nltk.download('punkt')# pre trained ML model for text tokenization
```

```python
df=pd.read_csv(r"C:\Users\shreyaash mogaveera\sentiment-analysis-nlp\data\IMDB Dataset.csv")
```

```python
#encoding target variable
df['label']=(df['sentiment']=='positive').astype(int)
df['label'].head()
```

```python
#text cleaning function
def clean_text(text):
    text=re.sub(r'<.*?>','',text) #remove html tags
    text=re.sub(r'http\S+|www\S+','',text) #remove urls
    text=re.sub(r'[^a-zA-Z\s]','',text)#remove special chars
    text=text.lower().strip() #lowercase
    text=re.sub(r'\s+',' ',text) #remove extra spaces
    return text
```

```python
df['cleaned_review']=df['review'].apply(clean_text)
df['cleaned_review'][0]
```

```python
#stopword removal
stop_words=set(stopwords.words('english'))
negation_words={'not','no','nor','neither','never',"don't","doesn't","didn't"}
stop_words=stop_words-negation_words

def remove_stopwords(tokens):
    return [w for w in tokens if w not in stop_words]
```

```python
#lemmatization
lemmatizer=WordNetLemmatizer()

def lemmatize_tokens(tokens):
    return [lemmatizer.lemmatize(t) for t in tokens]
```

```python
# applying all functions
df['tokens']=df['cleaned_review'].apply(str.split)
df['tokens']=df['tokens'].apply(remove_stopwords)
df['tokens']=df['tokens'].apply(lemmatize_tokens)
df['processed_review']=df['tokens'].apply(' '.join)
```
